```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# Step 1: PostgreSQL connection details
username = "postgres"
password = "7866"  # Your actual PostgreSQL password
host = "localhost"
port = 5432
database = "Pharma_Sales_Analysis"

# Step 2: Create database connection
connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

# Step 3: Load your DataFrame into PostgreSQL
table_name = "customer"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False
)

print(f"Data successfully loaded into table '{table_name}' "
      f"in database '{database}'.")

# Step 4: Close the connection
```
